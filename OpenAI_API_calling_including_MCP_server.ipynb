## Call the API Using the OpenAI Python Library (with Temperature)

```python
import openai
import json

messages = [
    {"role": "user", "content": "Hello, who are you?"}
]

response = openai.chat.completions.create(
    model="gpt-4",
    messages=messages,
    temperature=0.7
)

# Print the entire response as JSON
print(json.dumps(response.model_dump(), indent=2))

# Select and print specific fields
print("\nSession ID:", response.id)
print("Usage:", response.usage)
print("Reply:", response.choices[0].message.content)
```

## Call the API Using the Requests Library (with Temperature)

```python
import requests
import json

url = "https://api.openai.com/v1/chat/completions"
headers = {
    "Authorization": f"Bearer {OPENAI_API_KEY}",
    "Content-Type": "application/json"
}
data = {
    "model": "gpt-4",
    "messages": [
        {"role": "user", "content": "Hello, who are you?"}
    ],
    "temperature": 0.7
}

response = requests.post(url, headers=headers, json=data)
response_json = response.json()

# Print the entire JSON response, nicely formatted
print(json.dumps(response_json, indent=2))

# Select and print specific fields
print("\nSession ID:", response_json["id"])
print("Usage:", response_json["usage"])
print("Reply:", response_json["choices"][0]["message"]["content"])
```

# **Conversation History, Token Limit, and Keeping Only the Last 20 Messages**

## Using OpenAI Python Library

```python
import openai
import json

openai.api_key = OPENAI_API_KEY

# Start with an empty history
messages = []

# User sends the first message
user_message_1 = "Hello, who are you?"
messages.append({"role": "user", "content": user_message_1})

# Send to OpenAI
response_1 = openai.chat.completions.create(
    model="gpt-4",
    messages=messages,
    temperature=0.7,
    max_tokens=256
)

# Add assistant's reply to history
assistant_reply_1 = response_1.choices[0].message.content
messages.append({"role": "assistant", "content": assistant_reply_1})

print("Assistant:", assistant_reply_1)

# User sends a follow-up
user_message_2 = "What can you do?"
messages.append({"role": "user", "content": user_message_2})

# Keep only the last 20 messages
messages = messages[-20:]

# Send to OpenAI again
response_2 = openai.chat.completions.create(
    model="gpt-4",
    messages=messages,
    temperature=0.7,
    max_tokens=256
)

# Add assistant's reply to history
assistant_reply_2 = response_2.choices[0].message.content
messages.append({"role": "assistant", "content": assistant_reply_2})

print("Assistant:", assistant_reply_2)

# Show all assistant answers so far
assistant_answers = [msg["content"] for msg in messages if msg["role"] == "assistant"]
print("\nAll assistant answers so far:")
for i, ans in enumerate(assistant_answers, 1):
    print(f"{i}. {ans}")

# Show current conversation history length
print(f"\nCurrent conversation history length: {len(messages)} (max 20)")
```

## Create and Save Synthetic Sales Data

```python
import pandas as pd

# Create synthetic sales data
data = [
    [1001, "High-Waist Airlift Legging", "yoga pants", "HWAL-001", 110, 2],
    [1002, "5\" Airlift Energy Short", "shorts", "AE5S-002", 60, 1],
    [1003, "Sway Bra Tank", "tank tops", "SBT-003", 75, 1],
    [1004, "High-Waist Airlift Legging", "yoga pants", "HWAL-001", 110, 1],
    [1005, "5\" Airlift Energy Short", "shorts", "AE5S-002", 60, 3],
    [1006, "Sway Bra Tank", "tank tops", "SBT-003", 75, 2],
    [1007, "High-Waist Airlift Legging", "yoga pants", "HWAL-001", 110, 1],
    [1008, "5\" Airlift Energy Short", "shorts", "AE5S-002", 60, 2],
    [1009, "Sway Bra Tank", "tank tops", "SBT-003", 75, 1],
    [1010, "High-Waist Airlift Legging", "yoga pants", "HWAL-001", 110, 2],
]

df = pd.DataFrame(data, columns=["transaction_id", "product", "category", "sku", "price", "quantity"])
df.to_csv("sales_data.csv", index=False)
print("Synthetic sales_data.csv created!")
```

## Upload the CSV to OpenAI

```python
file_obj = openai.files.create(file=open("sales_data.csv", "rb"), purpose="assistants")
file_id = file_obj.id
print("Uploaded file ID:", file_id)
```

## Prepare the Responses API Payload

```python
import json

system_prompt = """
You are a pricing analyst for my clothing company. Please use the MCP tool
to fetch prices from the Alo Yoga MCP server for the categories of women's
shorts, yoga pants, and tank tops. Use only the MCP server for Alo yoga data, don't search the web.

Next, use the web search tool to search for Uniqlo prices for women's shorts, yoga pants, and tank tops.

In each case for Alo Yoga and Uniqlo, extract the
price for the top result in each category. Also provide the full URLs

Using the uploaded CSV file of sales data from my store, and with the
code interpreter tool calculate revenue by product item, compute average order-value on a transaction level, and calculate the percentage price gap between the CSV data and Uniqlo/Alo Yoga prices.
Flag products priced 15% or more above or below market.
Create and output a short report including the findings.
"""

payload = {
    "model": "gpt-4.1",
    "input": [
        {
            "role": "system",
            "content": [
                {"type": "input_text", "text": system_prompt}
            ]
        }
    ],
    "tools": [
        {
            "type": "web_search_preview",
            "user_location": {"type": "approximate", "country": "US"},
            "search_context_size": "medium"
        },
        {
            "type": "code_interpreter",
            "container": {
                "type": "auto",
                "file_ids": [file_id]
            }
        },
        {
            "type": "mcp",
            "server_url": "https://www.aloyoga.com/api/mcp",
            "server_label": "aloyoga",
            "allowed_tools": [
                "search_shop_catalog",
                "get_product_details"
            ],
            "require_approval": "never"
        }
    ],
    "temperature": 1,
    "max_output_tokens": 2048,
    "top_p": 1,
    "store": True
}
```

## Call the OpenAI Responses API

```python
import requests

api_key = OPENAI_API_KEY
url = "https://api.openai.com/v1/responses"

headers = {
    "Content-Type": "application/json",
    "Authorization": f"Bearer {api_key}"
}

response = requests.post(url, headers=headers, data=json.dumps(payload))
result = response.json()
print(json.dumps(result, indent=2))
```

## Display the report

```python
from IPython.display import Markdown, display

# Find the last assistant message in the output list
assistant_messages = [
    item for item in result['output']
    if item.get('type') == 'message' and item.get('role') == 'assistant'
]

if assistant_messages:
    last_message = assistant_messages[-1]
    # The report is in content[0]['text']
    report_md = last_message['content'][0]['text']
    display(Markdown(report_md))
else:
    print("No assistant message found in output.")
```
